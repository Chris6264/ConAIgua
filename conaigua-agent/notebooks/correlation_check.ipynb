# Comprobación de correlación
### Ejemplo: Correlación Pearson y Spearman entre precipitación y temperatura máxima en la estación 25001.

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_parquet('../data/processed/conAIgua_dataframe.parquet')
```

```python
df_f = df[df['estacion_id'] == '25001'].dropna(subset=['precip', 'tmax'])
```

```python
pearson_r, pearson_p = stats.pearsonr(df_f['precip'], df_f['tmax'])
spearman_r, spearman_p = stats.spearmanr(df_f['precip'], df_f['tmax'])
```

```python
print(f"Pearson  r={pearson_r:.4f}  p={pearson_p:.4f}  sig={pearson_p < 0.05}")
print(f"Spearman r={spearman_r:.4f}  p={spearman_p:.4f}  sig={spearman_p < 0.05}")
```
