```python
#Cargamos dataframe
df = pd.read_parquet('../data/processed/conAIgua_dataframe.parquet')
```

```python
#Promedio de precipitacion de la estacion 25001
df_f = df[(df['estacion_id'] == '25001')]
print(df_f['precip'].mean())
print(df_f['precip'].describe())
```

```python
#Correlacion entre precipitacion y mes de la estacion 25001 en el año 1978
df_f = df[(df['estacion_id'] == '25001') & (df['anio'] == 1978)]
corr = df_f[['precip', 'mes']].corr().iloc[0, 1]
corr
```
